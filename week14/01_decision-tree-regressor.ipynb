# Example 1: Understanding Decision Tree Regression

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
```

## 1. Generate data (nonlinear function: sine)

```python
rng = np.random.RandomState(1337)

dataset_size = 1000

x = np.random.uniform(low=-np.pi, high=np.pi, size=(dataset_size, ))
y = np.sin(x) + 0.5 * np.random.randn(dataset_size)
```

```python
plt.scatter(x, y)
```

```python
x = x[:, None]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

# 2. Train regression models (compare depths 2 and 5)

```python
model_dict = {}
mse_dict = {}
for max_depth in [1, 3, 5]:
    model = DecisionTreeRegressor(max_depth=max_depth)
    model.fit(x_train, y_train)
    model_dict[max_depth] = model
    mse_dict[max_depth] = mean_squared_error(y_true=y_test, y_pred=model.predict(x_test))
```

## 3. Create test inputs for prediction and visualization

```python
x_vis = np.arange(-np.pi, np.pi, 0.01)[:, np.newaxis]
```

```python
# 4. Visualize the results
fig, ax = plt.subplots()
ax.scatter(x_test, y_test, s=20, label="Test samples", alpha=0.5, color='tab:gray')
for max_depth, model in model_dict.items():
    y_pred = model.predict(x_vis)
    mse = mse_dict[max_depth]
    ax.plot(x_vis, y_pred, label=rf'{max_depth=} $\to$ Test MSE={mse:.3f}', linewidth=2)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
```

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from IPython.display import Image
```

```python
def create_dummy_image(height: int, width: int, batch_size: int = 1, num_channels: int = 1):
    """Function to create a dummy input image tensor
    Args:
        height, width: dimensions of the image
        batch_size: number of images in a batch
        num_channels: number of input channels
    Returns:
        A random tensor of shape (batch_size, num_channels, height, width)
    """
    return torch.randn(batch_size, num_channels, height, width)
```

# Case 1: No padding, no strides
- Input size: 4x4
- Kernel size: 3x3
- Stride = 1
- Padding = 0
- Output size: (4 - 3 + 0*2)//1 + 1 = 2 -> 2x2

![](https://github.com/slowmoyang/conv_arithmetic/raw/master/gif/no_padding_no_strides.gif)

```python
conv = nn.Conv2d(in_channels=num_channels, out_channels=1, kernel_size=3, stride=1, padding=0)

x = create_dummy_image(4, 4)
# output feature map
out = conv(x)
print(f'{out.shape=}')
```

## Case 2: No padding, strides

- Input size: 5x5
- Kernel size: 3x3
- Stride: 2
- Padding: 0
- Output size: floor((5 - 3)/2 + 1) = 2 -> 2x2


![](https://github.com/slowmoyang/conv_arithmetic/raw/master/gif/no_padding_strides.gif)

```python
conv = nn.Conv2d(in_channels=num_channels, out_channels=1, kernel_size=3, stride=2, padding=0)

x = create_dummy_image(5, 5)
out = conv(x)
print(f'{out.shape=}')
```

## Padding, strides
- Input size: 5x5
- Kernel: 3x3
- Padding: 1
- Stride: 2
- Output size: floor((5 + 2*1 - 3)/2 + 1) = 3 -> 3x3


![](https://github.com/slowmoyang/conv_arithmetic/raw/master/gif/padding_strides.gif)

```python
conv = nn.Conv2d(in_channels=num_channels, out_channels=1, kernel_size=3, stride=2, padding=1)

x = create_dummy_image(5, 5)
out = conv(x)
print(f'{out.shape=}')
```

## Case 4: No padding, no stride, dilation

- Dilation expands the kernel: effective kernel size = (3 - 1) * 2 + 1 = 5
- Input size: 7x7
- Output size: (7 - 5 + 0*2)//1 + 1 = 3 -> 3x3

![](https://github.com/slowmoyang/conv_arithmetic/raw/master/gif/dilation.gif)

```python
conv = nn.Conv2d(in_channels=num_channels, out_channels=1, kernel_size=3, stride=1, padding=0, dilation=2)

x = create_dummy_image(7, 7)
out = conv(x)
print(f'{out.shape=}')
```

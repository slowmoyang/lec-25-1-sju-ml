# Example 0: Convolution with the Handcrafted Kernel

```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import display
```

Load and display the input image

```python
image = Image.open('./data/valve.png')
display(image)
```

Convert the image to grayscale (single channel)

```python
image = image.convert('L')
display(image)
```

Convert the PIL image to a NumPy array and then to a PyTorch tensor

```python
image = np.array(image)
image = torch.from_numpy(image)
```

Add batch and channel dimensions: (H, W) → (1, 1, H, W)

```python
print(image.shape)
image = image.unsqueeze(0).unsqueeze(0)
print(image.shape)
```

Convert the image to float and normalize pixel values to [0, 1]

```python
image = image.float()
image /= 255
```

Define a Sobel kernel for vertical edge detection (see: https://en.wikipedia.org/wiki/Sobel_operator)

```python
kernel = torch.tensor(
    data=[[[
        [-1., 0., 1.],
        [-2., 0., 2.],
        [-1., 0., 1.]
    ]]],
    dtype=torch.float32
)
```

Apply 2D convolution with padding=1 to keep the output size same as input

```python
output = F.conv2d(image, kernel, padding=1)
print(f'{output=}')
```

Plot the input image, the kernel, and the output (edge-detected image)

```python
fig, axarr = plt.subplots(figsize=(15, 5), ncols=3)

axarr[0].imshow(image[0, 0], cmap='gray')
axarr[0].set_title("Input Image")

axarr[1].imshow(kernel[0, 0], cmap='gray')
axarr[1].set_title('Convolution Kernel')

axarr[2].imshow(output[0, 0], cmap='gray')
axarr[2].set_title("Edge Detected")


for ax in axarr:
    ax.axis('off')
```
